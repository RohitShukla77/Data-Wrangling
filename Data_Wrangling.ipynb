```python
class_a=pd.DataFrame({
    "Student_Id":[1,2],
    "Name":["Aman","Riya"],
    "Marks":[76,87]
})
class_b=pd.DataFrame({
    "Student_Id":[101,102],
    "Name":["Kabir","Neha"],
    "Marks":[82,84]
})
```

```python
result=pd.concat([class_a,class_b],axis=1)
print(result)
```

Merge()

```python
student_info=pd.DataFrame({
    "Student_ID":[1,2,3,4],
    "Name":["Aman","Riya","Kabir","Neha"],
    "Age":[14,13,14,15]
})
student_marks=pd.DataFrame({
    "Student_ID":[1,2,3,4],
    "Subject":["Math","English","Science","Hindi"],
    "Marks":[82,84,76,98]
})
```

```python
merged=pd.merge(student_info,student_marks,on="Student_ID",how="inner")
print(merged)
merged=pd.merge(student_info,student_marks,on="Student_ID",how="left")
print(merged)
merged=pd.merge(student_info,student_marks,on="Student_ID",how="right")
print(merged)
merged=pd.merge(student_info,student_marks,on="Student_ID",how="outer")
print(merged)
```

Merging when key-column names differ

```python
student_info=pd.DataFrame({
    "Student_ID":[1,2,3,4],
    "Name":["Aman","Riya","Kabir","Neha"],
    "Age":[14,13,14,15]
})
student_marks=pd.DataFrame({
    "ID":[1,2,3,4],
    "Subject":["Math","English","Science","Hindi"],
    "Marks":[82,84,76,98]
})
ans=pd.merge(student_info,student_marks,left_on="Student_ID",right_on="ID",how="inner")
print(ans)
```

Join by index

```python
left=pd.DataFrame({
    "A":[1,2,3,4],
    "B":[5,6,7,8]
})
right=pd.DataFrame({
    "C":[9,10,11,12],
    "D":[13,14,15,16]
})
result=left.join(right)
print(result)
```

Reshaping and Pivoting

```python
wide=pd.DataFrame({
    "Student":['Aman','Riya','Kabir','Ram'],
    "Math":[82,84,76,98],
    "English":[82,84,76,98]
})
print(wide)
```

Convert wide data to Long using melt()

```python
long=pd.melt(wide,
             id_vars=['Student'],
             var_name='Subject',
             value_name='Marks')
print(long)
```

Pivot table()

```python
sales=pd.DataFrame({
    "Date":['2023-01-01','2023-01-01','2023-01-02','2023-01-02'],
    "Store":['A','B','A','B'],
    "Product":['X','Y','X','Y'],
    "Sales":[100,200,150,300]
})
summary=sales.pivot_table(
    values="Sales",
    index=["Store","Product"],
    aggfunc="sum"
)
print(summary)
```

Cross Tabulation

```python
data=pd.DataFrame({
    "Gender":['M','F','M','F','M'],
    "Age":[25,30,22,28,35]
})
cross=pd.crosstab(
    index=data["Gender"],
    columns=pd.cut(x=data["Age"],bins=[20,25,30,35])
)
print(cross)
```
